**Creating a Tuple**

```python

# empty tuple
my_tuple = ()

# tuple having integers
my_tuple = (1, 2, 3)

# tuple with mixed datatypes
my_tuple = (1, "Hello", 3.4)

# nested tuple
my_tuple = ("mouse", [8, 4, 6], (1, 2, 3))

# tuple can be created without parentheses
# also called tuple packing
my_tuple = 3, 4.6, "dog"
# tuple unpacking is also possible
a, b, c = my_tuple
```

```python

my_tuple = ['p','e','r','m','i','t']
print(my_tuple[0])

print(my_tuple[5])

print(my_tuple[6])   # index must be in range

#IndexError: list index out of range
print(my_tuple[2.0]) # index must be an integer


```

**Indexing**

```python
#TypeError: list indices must be integers, not float
n_tuple = ("mouse", [8, 4, 6], (1, 2, 3))
print(n_tuple[0][3])  # nested index

print(n_tuple[1][1])  # nested index

print(n_tuple[2][0])  # nested index
```

**Negative Indexing**

```python

 my_tuple = ['p','e','r','m','i','t']
 print(my_tuple[-1])
 print(my_tuple[-6])
```

**Slicing**

```python
 my_tuple = ('p','r','o','g','r','a','m','i','z')

print(my_tuple[1:4])  # elements 2nd to 4th

print(my_tuple[:-7])  # elements beginning to 2nd

print(my_tuple[7:])   # elements 8th to end

 print(my_tuple[:])   # elements beginning to end
```

**Changing or Deleting a Tuple**

```python
my_tuple = (4, 2, 3, [6, 5])
my_tuple[1] = 9 # we cannot change an element
my_tuple[3] = 9
```

```python
my_tuple[3][0] = 9   # but item of mutable element can be changed
my_tuple
```

```python

my_tuple = ('p','r','o','g','r','a','m','i','z') # tuples can be reassigned
my_tuple
```

```python

my_tuple = ('p','r','o','g','r','a','m','i','z')
del my_tuple[3] # can't delete items
```

```python

del my_tuple    # can delete entire tuple
my_tuple
```

**Python Tuple Methods**

```python
my_tuple = ('a','p','p','l','e',)
my_tuple.count('p')
```

```python
my_tuple.index('l')
```
